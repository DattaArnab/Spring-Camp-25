```python
import cv2
import os
import time
import pandas as pd
from datetime import datetime

# Configuration
output_dir = "val"
csv_file = "val.csv"
class_name = "FuckOff"  # Class name as a variable
new_label = 5 # Label for the new class
num_frames = 30  # Number of frames per video
capture_start = 0.5  # Start capturing video after 0.5 seconds
capture_end = 2.5  # Stop capturing video at 2.5 seconds
video_duration = 4  # Total duration of the video in seconds
repeat_count = 20 # Number of times to capture videos

# Ensure output directory exists
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# Load or create the CSV file
if os.path.exists(csv_file):
    df = pd.read_csv(csv_file, delimiter=';')
else:
    df = pd.DataFrame(columns=["Folder", "Gesture", "Label"])

# Function to capture video and save frames
def capture_video_and_save_frames(output_folder, label, count):
    cap = cv2.VideoCapture(0)  # Open webcam
    if not cap.isOpened():
        print("Error: Could not access the webcam.")
        return None

    print("Get ready...")
    time.sleep(1)

    for i in range(1, 4):  # Countdown: 1, 2, 3...
        print(i)
        time.sleep(1)

    print("Go!")
    start_time = time.time()
    frames = []

    while True:
        ret, frame = cap.read()
        if not ret:
            print("Error: Could not read frame.")
            break

        elapsed_time = time.time() - start_time

        # Capture frames only within the specified range (0.5s to 2.5s)
        if capture_start <= elapsed_time <= capture_end:
            frames.append(frame)

        # Stop capturing after the total duration
        if elapsed_time > video_duration:
            break

    cap.release()

    # Ensure we have exactly num_frames by sampling or padding
    if len(frames) < num_frames:
        frames.extend([frames[-1]] * (num_frames - len(frames)))  # Pad with the last frame
    elif len(frames) > num_frames:
        step = len(frames) // num_frames
        frames = frames[:num_frames * step:step]  # Sample evenly

    # Use current date and time for folder name
    timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
    folder_name = f"{output_folder}_{timestamp}"

    folder_path = os.path.join(output_dir, folder_name)
    os.makedirs(folder_path, exist_ok=True)

    for i, frame in enumerate(frames):
        frame_path = os.path.join(folder_path, f"frame_{i+1}.jpg")
        cv2.imwrite(frame_path, frame)

    return folder_name

# Main loop to capture videos and update CSV file
for count in range(1, repeat_count + 1):
    folder_name = capture_video_and_save_frames(class_name.replace(" ", "_"), new_label, count)

    if folder_name:
        print(f"Captured and saved video {count} as {folder_name}.")

        # Append new entry to DataFrame
        df.loc[len(df)] = [folder_name, class_name, new_label]

# Save updated CSV file
df.to_csv(csv_file, sep=';', index=False)
print(f"Updated CSV file saved as {csv_file}.")
```
